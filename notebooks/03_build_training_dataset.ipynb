# Auspex Dataset Construction

## Objective

Create a clean training dataset for Auspex using the raw CIC-Bell-DNS2021 domain lists.

The final dataset will contain:

- domain
- label

where label is one of:

- benign
- phishing
- malware
- spam

This dataset will be used for feature engineering and model training.

## Why We Chose the Raw Dataset

The CIC-Bell-DNS2021 processed CSVs contain engineered DNS features and parsing inconsistencies.

The raw domain lists:

- are cleaner
- directly match user input
- are easier to explain
- allow us to generate our own features


# Auspex Dataset Construction

## Objective

Create a clean training dataset from the CIC-Bell-DNS2021 raw domain lists.

Final schema:

| domain | label |
|---------|---------|
| google.com | benign |
| evil-login.net | phishing |

Labels:
- benign
- phishing
- malware
- spam

This dataset will be used for feature engineering and model training.

```python
import pandas as pd
from urllib.parse import urlparse
```

```python
benign = pd.read_csv(
    "../data/raw/cic_bell_dns2021_raw/benign_domains.csv",
    header=None,
    names=["domain"]
)

malware = pd.read_csv(
    "../data/raw/cic_bell_dns2021_raw/malware_domains.csv",
    header=None,
    names=["domain"]
)

spam = pd.read_csv(
    "../data/raw/cic_bell_dns2021_raw/spam_domains.csv",
    header=None,
    names=["domain"]
)

print("Benign:", benign.shape)
print("Malware:", malware.shape)
print("Spam:", spam.shape)
```

```python
phishing_raw = pd.read_csv(
    "../data/raw/cic_bell_dns2021_raw/phishing_domains.csv",
    header=None
)

print(phishing_raw.shape)

phishing_raw.head()
```

```python
phishing_raw.iloc[:10, :5]
```

```python
def extract_domain(url):
    try:
        return urlparse(url).netloc.lower()
    except:
        return None
```

```python
phishing = pd.DataFrame()

phishing["domain"] = phishing_raw[1].apply(extract_domain)

phishing.head()
```

```python
benign["label"] = "benign"
malware["label"] = "malware"
spam["label"] = "spam"
phishing["label"] = "phishing"
```

```python
combined = pd.concat(
    [
        benign,
        phishing,
        malware,
        spam
    ],
    ignore_index=True
)

print(combined.shape)
combined.head()
```

```python
combined = combined.dropna()

combined["domain"] = combined["domain"].str.strip()

combined = combined.drop_duplicates()

print(combined.shape)
```

```python
combined.groupby("domain")["label"].nunique().sort_values(ascending=False)
```

```python
conflicts = combined.groupby("domain")["label"].nunique()

conflicts = conflicts[conflicts > 1]

print(len(conflicts))
```

```python
conflicting_domains = conflicts.index

combined[
    combined["domain"].isin(conflicting_domains)
].sort_values("domain")
```

Domains with conflicting labels were removed.

Reason:
The same domain occasionally appeared under multiple labels
(e.g. benign and phishing).

To reduce label ambiguity, all conflicting domains were excluded
from the final training dataset.

```python
conflicts = (
    combined.groupby("domain")["label"]
    .nunique()
)

bad_domains = conflicts[
    conflicts > 1
].index

cleaned = combined[
    ~combined["domain"].isin(bad_domains)
]
```

```python
cleaned.shape
```

```python
print(cleaned.shape)

cleaned.to_csv(
    "../data/processed/auspex_domains_v1.csv",
    index=False
)

print("Saved auspex_domains_v1.csv")
```
